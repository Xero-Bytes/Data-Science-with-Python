# 2D Array

**2D Array Consists of Rows and Colums**

```python
a=np.array([
    [10,20,30],
    [40,50,60]
])
print(a)
print(a.ndim)
```

# 3D Array

**Multiple 2D Array Collections**

```python
a=np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
           ])
print(a)
print(a.ndim)
```

**Important Array Attributes**

```python
a=np.array([
    [10,20,30],
    [40,50,60]
])
print(a.ndim)  #number of dimensions
print(a.shape) #Rows,Columns#
print(a.size)  #Total elements(total Values)
print(a.dtype) #data Types
```

# Creating Arrays
**Multiple Methods To Create Numpy Array**

## 1. np.array()

```python
a=np.array([1,2,3,4])
print(a)
```

## 2. np.zeros()

```python
a=np.zeros(5)
print(a)
```

```python
a=np.zeros((2,3))
print(a)
```

# 3. np.ones()

```python
a=np.ones((2,3))
print(a)
```

# 4. np.full()
** Fill array with specific values

```python
a=np.full((2,3),7)
print(a)
```

# 5. np.arrange()

```python
a=np.arange(1,11)
print(a)
```

```python
a=np.arange(2,11,2) #start , Stop , step
print(a)
```

# 6. np.linspace()
equally spaced numbers b/w two values

# 7. np.eye()

```python
a=np.eye(4)
print(a)
```

## Indexing And Slicing

# 1D Indexing
negative indexing start from end

```python
a=np.array([10,20,30,40,50])
print(a[0])
print(a[1])
print(a[2])
print(a[-1])
print(a[-2])
print(a[3])
print(a[4])
```

# Slicing
***select range values***

```python
a=np.array([10,20,30,40,50,60])
print(a[1:5])  #last not include
```

```python
print(a[:3])  #First 3
print(a[2:])  #index 2 onward
print(a[::2]) #every Second
print(a[::-1]) #Reverse
```

# 2D Indexing

```python
a=np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(a[0,1])
```

```python
print(a[1,:])
```

```python
print(a[:,1])
```

```python
print(a[0:2,1:3])
```

# Mathematical Calculation

```python
a=np.array([4,9,16])
print(np.sqrt(a))
```

```python
a=np.array([1.2,2.7,3.5])
print(np.floor(a))
print(np.ceil(a))
```
