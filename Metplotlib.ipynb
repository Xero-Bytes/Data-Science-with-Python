# MetplotLib

**1. Figure**

A figure is the complete canvas or window on which a graph is created.

**2. Axes**

Axes are the actual plotting areas inside a figure. A figure can contain one or more plotting areas.

**3. X-axis**

The X-axis is the horizontal axis. It usually represents independent values, such as days, months, or student names.

**4. Y-axis**

The Y-axis is the vertical axis. It usually represents values such as marks, sales, temperature, or prices.

**5. Title**

The title describes what the graph represents.

**6. Labels**

Labels explain what the X-axis and Y-axis represent.

```python
import matplotlib.pyplot as plt
```

```python
days=[1,2,3,4,5]
temperature=[25,26,27,28,29]
plt.plot(days,temperature)     #create the line graph
plt.show()
```

```python
plt.plot(days,temperature,marker="o",color="green")  # displays a circle at each data point.  changes the line color to green.
plt.title("Weekly Temperature")   #adds a title to the graph.
plt.xlabel("Day")                 #adds a label to the X-axis.
plt.ylabel("Temperature (C*)")    # adds a label to the Y-axis.
plt.grid(True)                    #displays grid lines.
plt.show()
```

# Types of Graphs in Matplotlib
***Matplotlib provides different types of graphs. Each graph is suitable for a particular purpose.***

# A. Line Plot

A line plot is used to show changes or trends over time.

***Examples:***
1- Monthly sales
2- Daily temperature
3- Stock prices
4- Student performance over several tests

***Function: plt.plot()***

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 130, 200]
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

**The line connects the data points and shows how sales change from month to month.**

# B. Bar Chart

A bar chart is used to compare different categories.

**Examples:**
1- Students' marks
2- Product sales
3- Employee numbers
4- Department performance

***Function: plt.bar()***

```python
import matplotlib.pyplot as plt
students = ["Ali", "Sara", "Ahmed"]
marks = [70, 90, 80]
plt.bar(students, marks)
plt.title("Students' Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

**Each bar represents one student, and the height of the bar represents their marks.
A horizontal bar chart can be created using plt.barh().**

```python
import matplotlib.pyplot as plt
students = ["Ali", "Sara", "Ahmed"]
marks = [70, 90, 80]
plt.barh(students, marks)
plt.title("Students' Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

# C. Scatter Plot

A scatter plot displays individual data points. It is useful for identifying relationships or patterns between two numerical variables.

**Examples:**
1- Study hours versus marks
2- Advertising costs versus sales
3- Height versus weight
4- xperience versus salary

****Function: plt.scatter()****

```python
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5]
marks = [40, 50, 65, 75, 90]
plt.scatter(hours,marks)
plt.title("Study Hours and Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

**Each dot represents a pair of values. The pattern of dots can help us understand how the variables are related.**

# D. Histogram

A histogram shows the distribution of numerical data. It groups values into intervals called bins.

**Examples:**

1- Distribution of exam marks
2- Distribution of ages
3- Distribution of salaries
4- Distribution of product prices

**Function: plt.hist()**

```python
import matplotlib.pyplot as plt
marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85]
plt.hist(marks,bins=5,edgecolor="black")                       #bins=5 divides the data range into five intervals.
plt.title("Distribution of marks")
plt.xlabel("Marks")
plt.ylabel("number of Students")
plt.show()
```

**The height of each bar represents the number of observations in that interval.**

# E. Pie Chart

A pie chart represents how different categories contribute to a whole.

**Examples:**
    
1- Monthly expenses
2- Company market share
3- Budget distribution
4- Percentage of students in different departments

**Function: plt.pie()**

```python
import matplotlib.pyplot as plt
categories = ["Food", "Travel", "Books", "Other"]
expenses = [40, 20, 15, 25]
plt.pie(
    expenses,
    labels=categories,     #labels displays the category names.
    autopct="%1.1f%%"      #autopct="%1.1f%%" displays percentages with one decimal place.
)
plt.title("Monthly Expenses")
plt.show()
```

**Pie charts work best when the categories represent parts of the same total.**

# Changing Figure Size

Sometimes a graph is too small or its labels are difficult to read. We can change its size using plt.figure().

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))      # sets the figure width to 8 inches and height to 5 inches.
plt.plot(
    [1, 2, 3, 4],
    [10, 20, 15, 30],
    marker="o",
    label="Sales"                # assigns a name to the plotted data.
)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()                    # displays the label on the graph.
plt.show()
```

# Saving a Graph as an Image

Matplotlib allows us to save graphs in formats such as PNG, PDF, and SVG.

```python
import matplotlib.pyplot as plt
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("My First Graph")
plt.savefig(                            #saves the graph.
    "my_graph.png",                     #is the file name.
    dpi=150,                            #sets the image resolution.
    bbox_inches="tight"                 #reduces unnecessary blank space around the graph.
)
plt.show()
```

**It is good practice to call savefig() before show().**

The file is saved in the current working directory unless you specify another path.

# Using Matplotlib with Pandas

Pandas is a Python library used to organize and analyze data. It can work with Matplotlib to create graphs directly from a DataFrame.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({                       #creates a DataFrame.
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [100, 150, 130, 200]
})
df.plot(
    x="Month",                            #selects the column for the X-axis.
    y="Sales",                            #selects the values to plot.
    kind="bar"                            #specifies a bar chart.
)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({                       #creates a DataFrame.
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [100, 150, 130, 200]
})
df.plot(
    x="Month",                            #selects the column for the X-axis.
    y="Sales",                            #selects the values to plot.
    kind="hist"                            #specifies a histogram chart.
)
plt.show()
```

# Matplotlib and Machine Learning

Matplotlib is especially useful in machine learning because it helps us understand data and model performance.

Common applications include:

**- Visualizing training and validation loss.**
    
**- Comparing actual values with predicted values.**
    
**- Understanding data distributions.**
    
**- Identifying relationships between features.**
    
**- Plotting confusion matrices.**
    
**- Visualizing model accuracy over time.**
    
For example, a machine learning model's loss over several epochs can be plotted using a line graph:

```python
import matplotlib.pyplot as plt
epochs = [1, 2, 3, 4, 5]
loss = [0.90, 0.70, 0.50, 0.35, 0.25]
plt.plot(epochs, loss, marker="o")
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```
